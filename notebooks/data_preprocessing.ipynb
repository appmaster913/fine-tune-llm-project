# Data Preprocessing

This notebook contains the code for preprocessing the raw customer data used for fine-tuning the language model. The preprocessing steps include data cleaning, transformation, and preparation for training.

```python
import pandas as pd
import numpy as np

# Load raw data
raw_data_path = '../data/raw/customer_data.csv'
data = pd.read_csv(raw_data_path)

# Display the first few rows of the dataset
data.head()
```

```python
# Data cleaning steps

# Remove duplicates
data = data.drop_duplicates()

# Handle missing values
data.fillna(method='ffill', inplace=True)

# Convert categorical columns to numeric
data['category'] = data['category'].astype('category').cat.codes

# Display cleaned data
data.info()
```

```python
# Data transformation steps

# Example transformation: Normalizing a numeric column
data['normalized_column'] = (data['numeric_column'] - data['numeric_column'].mean()) / data['numeric_column'].std()

# Save processed data
processed_data_path = '../data/processed/cleaned_customer_data.csv'
data.to_csv(processed_data_path, index=False)

# Display a message indicating completion
print('Data preprocessing complete. Processed data saved to:', processed_data_path)
```
