# Model Evaluation

This notebook contains code for evaluating the performance of the fine-tuned language model. We will use various metrics to assess the model's effectiveness on the validation dataset.

```python
import json
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from scripts.evaluate_model import evaluate_model

# Load evaluation configuration
with open('../configs/evaluation_config.json') as config_file:
    eval_config = json.load(config_file)

# Load the fine-tuned model
model = ...  # Load your fine-tuned model here

# Load validation data
validation_data = ...  # Load your validation dataset here

# Evaluate the model
predictions, true_labels = evaluate_model(model, validation_data)

# Calculate metrics
accuracy = accuracy_score(true_labels, predictions)
precision = precision_score(true_labels, predictions, average='weighted')
recall = recall_score(true_labels, predictions, average='weighted')
f1 = f1_score(true_labels, predictions, average='weighted')

# Display metrics
metrics = {
    'Accuracy': accuracy,
    'Precision': precision,
    'Recall': recall,
    'F1 Score': f1
}

for metric, value in metrics.items():
    print(f"{metric}: {value:.4f}")

# Visualize results
labels = list(metrics.keys())
values = list(metrics.values())

plt.bar(labels, values, color=['blue', 'orange', 'green', 'red'])
plt.ylim(0, 1)
plt.title('Model Evaluation Metrics')
plt.ylabel('Score')
plt.show()
```
